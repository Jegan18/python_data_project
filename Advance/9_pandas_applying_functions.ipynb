```python
#Importing libraries
import pandas as pd
from datasets import load_dataset

#loading data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#Data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_salary = df[pd.notna(df['salary_year_avg'])]['salary_year_avg']
```

```python
df_salary = df[pd.notna(df['salary_year_avg'])].copy() #whenever you change something original df you have to copy this

def projected_salary(salary):
    return salary * 1.03 #3% inflation

df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)

df_salary[['salary_year_avg','salary_year_inflated']] #comparing 
```

```python
df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)

df_salary[['salary_year_avg','salary_year_inflated']] #comparing  lambda
```
