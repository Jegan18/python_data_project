```python
df_US_software_pivot = pd.read_csv('https://lukeb.co/software_csv', index_col='job_posted_month')

df_US_software_pivot
```

```python
import nbformat as nbf
import pandas as pd
import pandas as pd

nb = nbf.v4.new_notebook()

nb.cells = [
    nbf.v4.new_code_cell(
        """

# Use the pivot table already available in this notebook
if 'df_US_software_pivot' not in globals():
    raise NameError('df_US_software_pivot is not defined in the current notebook.')

# Create an alias that matches your requested name
df_US_pivot = df_US_software_pivot.reset_index().copy()
df_US_pivot = df_US_pivot.rename(columns={'job_posted_month': 'month'})

# Small mapping table to demonstrate a merge
df_role_map = pd.DataFrame({
    'role': [
        'Front-End Developer',
        'Back-End Developer',
        'Full-Stack Developer',
        'UI/UX Designer'
    ],
    'team': ['Frontend', 'Backend', 'Fullstack', 'Design']
})

df_US_pivot.head()
"""
    ),
    nbf.v4.new_code_cell(
        """

# Convert the wide pivot table to long format for merging
df_US_pivot_long = df_US_pivot.melt(
    id_vars='month',
    var_name='role',
    value_name='job_count'
)

# Merge with the role mapping table
df_US_pivot_merge = df_US_pivot_long.merge(df_role_map, on='role', how='left')

df_US_pivot_merge.head()
"""
    )
]

with open('6_exercise_job_demand.ipynb', 'w', encoding='utf-8') as f:
    nbf.write(nb, f)

print('Created 6_exercise_job_demand.ipynb')
```
