```python
#Importing libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#loading data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#Data cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df=df[df['job_title_short'] == 'Data Analyst']
```

```python
df_explode = df.explode('job_skills')

skill_stats = df_explode.groupby('job_skills').agg(
    skills_count = ('job_skills','count'),
    median_salary = ('salary_year_avg','median')
)


skill_stats = skill_stats.sort_values(by='skills_count',ascending=False).head(20)
```

```python
from adjustText import adjust_text

# make sure the cell that defines `skill_stats` has been run first
skill_stats.plot(kind='scatter', x='skills_count', y='median_salary')

texts = []
for i, txt in enumerate(skill_stats.index):
    texts.append(plt.text(
        skill_stats['skills_count'].iloc[i],
        skill_stats['median_salary'].iloc[i],
        txt
    ))
adjust_text(texts)

plt.xlabel('Count of Job Posting')
plt.ylabel('Median salary yearly')
plt.title('Salary vs Count of top 10 skills')
plt.tight_layout()

plt.show()


texts = []
for i, txt in enumerate(skill_stats.index):
  texts.append(plt.text(
        skill_stats['skills_count'].iloc[i],
        skill_stats['median_salary'].iloc[i],
        txt
    ))
adjust_text(texts)

plt.xlabel('Count of Job Posting')
plt.ylabel('Median salary yearly')
plt.title('Salary vs Count of top 10 skills')
plt.tight_layout()

plt.show()
```
